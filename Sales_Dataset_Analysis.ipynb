NumPy is the primary library used in this project for numerical analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the Dataset

The sales dataset is stored as a CSV file.

We first load the dataset using Pandas so that we can inspect its structure and then extract numerical columns for NumPy analysis.

```python
af = pd.read_csv("Sales Dataset.csv")
```

## 3. Identify Numerical Variables

For the NumPy analysis, we will focus on the numerical variables in the dataset.

The main numerical variables are:

- Amount
- Profit
- Quantity

These variables allow us to perform numerical calculations using NumPy arrays.

```python
af[["Amount", "Profit", "Quantity"]].head()
```

```python
amount = af["Amount"].to_numpy()
profit = af["Profit"].to_numpy()
quantity = af["Quantity"].to_numpy()

print("Amount:", amount[:10])
print("Profit:", profit[:10])
print("Quantity:", quantity[:10])
```

## 4. Understanding NumPy Arrays

After extracting the numerical columns from the DataFrame, we can examine the structure of the resulting NumPy arrays.

We will inspect:

- Number of observations
- Number of dimensions
- Shape
- Data type

```python
print("Amount")
print("Size:", amount.size)
print("Dimensions:", amount.ndim)
print("Shape:", amount.shape)
print("Data type:", amount.dtype)
```

```python
print("\nProfit")
print("Size:", profit.size)
print("Dimensions:", profit.ndim)
print("Shape:", profit.shape)
print("Data type:", profit.dtype)

print("\nQuantity")
print("Size:", quantity.size)
print("Dimensions:", quantity.ndim)
print("Shape:", quantity.shape)
print("Data type:", quantity.dtype)
```

## 5. Total Sales

The total sales amount represents the combined sales value across all transactions.

NumPy's `sum()` function can be used to calculate the total sales.

```python
total_sales = np.sum(amount)

print("Total Profit" , total_sales)
```

```python
total_profit = np.sum(profit)

print("Total Profit" , total_profit)
```

```python
total_quantity = np.sum(quantity)

print("Total Quantity" , total_quantity)
```

```python
average_sales = np.mean(amount)

print("Average Sales" ,average_sales)
```

```python
average_profit = np.mean(profit)

print("Average Profit" , average_profit)
```

```python
average_quantity = np.mean(quantity)

print("Average Quantity", average_quantity)
```

```python
print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Quantity:", total_quantity)

print("Average Sales:", average_sales)
print("Average Profit:", average_profit)
print("Average Quantity:", average_quantity)
```

## Insight

The dataset contains a total sales value of 6,182,639 and a total profit of 1,610,697 across 12,745 units sold.

On average, each transaction generated approximately:

- Sales Amount: 5,178.09
- Profit: 1,348.99
- Quantity Sold: 10.67 units

These metrics provide an initial understanding of the business performance and establish a baseline for further analysis of sales and profitability.

```python
min_sales = np.min(amount)
max_sales = np.max(amount)

print("Minimum Sales:", min_sales)
print("Maximum Sales:", max_sales)
```

```python
min_profit = np.min(profit)
max_profit = np.max(profit)

print("Minimum Profit:", min_profit)
print("Maximum Profit:", max_profit)
```

```python
min_quantity = np.min(quantity)
max_quantity = np.max(quantity)

print("Minimum Quantity:", min_quantity)
print("Maximum Quantity:", max_quantity)
```

```python
sales_max_index = np.argmax(amount)
sales_min_index = np.argmin(amount)

print("Highest Sales Position:", sales_max_index)
print("Lowest Sales Position:", sales_min_index)
```

```python
af.iloc[sales_max_index]
```

```python
np.min(amount)
np.max(amount)

np.min(profit)
np.max(profit)

np.min(quantity)
np.max(quantity)
```

```python
sales_max_index = np.argmax(amount)
sales_min_index = np.argmin(amount)

af.iloc[sales_max_index]
af.iloc[sales_min_index]
```

### Insight

The minimum sales transaction generated sales of 508, while the corresponding transaction generated a profit of 177.

The transaction involved 17 units of a Furniture product from the Sofas sub-category.

This demonstrates how `np.argmin()` can be used to identify the position of the minimum value in a NumPy array, after which the corresponding record can be retrieved from the original DataFrame using `.iloc[]`.
